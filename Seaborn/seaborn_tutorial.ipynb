```python
import matplotlib.pyplot as plt 
import seaborn as sns
fig, ax = plt.subplots(figsize=(7,5))
 
data=sns.load_dataset("tips")
sns.scatterplot(
   data=data,
    x="total_bill",
    y="tip", 
    hue="day",
    ax=ax
) 
sns.set_style("whitegrid") 
ax.set_title("Scatter Plot with Seaborn and Matplotlib")
plt.show()
```

```python
tips=sns.load_dataset("tips")
fig,ax=plt.subplots(figsize=(7,3))
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)
ax.set_title("Total Bill vs Tip")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset("tips")
fig,ax=plt.subplots(figsize=(12,5))
sns.boxplot(
    data=tips,
    x="total_bill",
    y="day",
    hue="sex",
    ax=ax
)
ax.set_title("total_bill vs day")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset("tips")
sns.set_style("whitegrid")
fig,ax=plt.subplots(figsize=(8,5))
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    ax=ax
)
ax.set_title("total_bill vs tip")
plt.show()
```
